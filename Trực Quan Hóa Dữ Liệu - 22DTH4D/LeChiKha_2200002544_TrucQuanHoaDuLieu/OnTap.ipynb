**Bài 1:** **Column Chart – Doanh thu theo tháng:**

**Yêu cầu:**

Vẽ biểu đồ cột thể hiện doanh thu của một cửa hàng theo từng tháng trong năm.

**Dữ liệu mẫu:**

```python
import matplotlib.pyplot as plt
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [120, 150, 170, 130, 190, 210]

plt.bar(months, sales)
plt.xlabel('Tháng')
plt.ylabel('Doanh thu')
plt.title('Doanh thu theo tháng')
plt.show()
```

**Bài 2: Radar Chart – Kỹ năng cá nhân:**

**Yêu cầu:**

Vẽ biểu đồ radar thể hiện điểm số của một học sinh về 6 kỹ năng: Toán, Lý, Hóa, Văn, Anh, Tin.

**Dữ liệu mẫu:**

labels = ['Toán', 'Lý', 'Hóa', 'Văn', 'Anh', 'Tin']

scores = [85, 90, 75, 70, 80, 95]


```python
import matplotlib.pyplot as plt
import numpy as np

labels = ['Toán', 'Lý', 'Hóa', 'Văn', 'Anh', 'Tin']
scores = [85, 90, 75, 70, 80, 95]

angles = np.linspace(0, 2 * np.pi, len(labels), endpoint=False).tolist()
angles += angles[:1]

fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
ax.plot(angles, scores + scores[:1], linewidth=1, linestyle='solid')
ax.fill(angles, scores + scores[:1], alpha=0.25)
ax.set_thetagrids(np.degrees(angles[:-1]), labels)
ax.set_ylim(0, 100)
plt.title('Kỹ năng cá nhân')
plt.show()

```

# **Ôn Tập 2:**

**Bài 1. Pyramid Chart (Biểu đồ kim tự tháp)**

**Đề bài:** Viết chương trình vẽ biểu đồ kim tự tháp dân số cho một quốc gia với dữ liệu dân số nam và nữ theo các nhóm tuổi.

**Dữ liệu:**

age_groups = ['0-4', '5-9', '10-14', '15-19', '20-24', '25-29']

male_population = [5.5, 5.3, 5.1, 5.0, 4.9, 4.8]

female_population = [5.2, 5.1, 4.9, 4.8, 4.7, 4.6]


```python
import matplotlib.pyplot as plt

# Dữ liệu
age_groups = ['0-4', '5-9', '10-14', '15-19', '20-24', '25-29']
male_population = [5.5, 5.3, 5.1, 5.0, 4.9, 4.8]
female_population = [5.2, 5.1, 4.9, 4.8, 4.7, 4.6]

# Đảo ngược danh sách để vẽ từ trẻ đến già theo trục y
age_groups = age_groups[::-1]
male_population = male_population[::-1]
female_population = female_population[::-1]

# Thiết lập biểu đồ
fig, ax = plt.subplots(figsize=(10, 6))

# Vẽ biểu đồ ngang: Nam bên trái (số âm), Nữ bên phải
ax.barh(age_groups, [-x for x in male_population], color='blue', label='Nam')
ax.barh(age_groups, female_population, color='red', label='Nữ')

# Thêm tiêu đề và nhãn
ax.set_title('Biểu đồ Kim Tự Tháp Dân Số')
ax.set_xlabel('Dân số (triệu người)')
ax.set_ylabel('Nhóm tuổi')

# Định dạng trục x để hiện số dương
xticks = list(range(-6, 7, 1))
ax.set_xticks(xticks)
ax.set_xticklabels([abs(x) for x in xticks])

# Thêm đường lưới và chú thích
ax.grid(True, axis='x', linestyle='--', alpha=0.7)
ax.legend()

# Hiển thị biểu đồ
plt.tight_layout()
plt.show()
```

**Bài 2. Bubble Chart (Biểu đồ bong bóng)**

**Đề bài:** Viết chương trình thể hiện dữ liệu GDP, dân số và tuổi thọ của các quốc gia dưới dạng biểu đồ bong bóng.

**Dữ liệu:**

countries = ['Vietnam', 'USA', 'Japan']

gdp = [3000, 65000, 40000]

life_expectancy = [73, 79, 84]

population = [97, 331, 125]  # triệu người


```python
# prompt: Viết chương trình thể hiện dữ liệu GDP, dân số và tuổi thọ của các quốc gia dưới dạng biểu đồ bong bóng.

import matplotlib.pyplot as plt

# Dữ liệu
countries = ['Vietnam', 'USA', 'Japan']
gdp = [3000, 65000, 40000]
life_expectancy = [73, 79, 84]
population = [97, 331, 125]  # triệu người

# Vẽ biểu đồ bong bóng
plt.figure(figsize=(10, 6))
plt.scatter(gdp, life_expectancy, s=population, alpha=0.6, c=population, cmap='viridis')

# Thêm chú thích cho mỗi bong bóng
for i, country in enumerate(countries):
    plt.annotate(country, (gdp[i], life_expectancy[i]))

# Thêm tiêu đề và nhãn
plt.title('Biểu đồ Bong bóng GDP, Tuổi thọ và Dân số')
plt.xlabel('GDP (tỷ USD)')
plt.ylabel('Tuổi thọ trung bình')

# Thêm thanh màu
plt.colorbar(label='Dân số (triệu người)')

# Hiển thị biểu đồ
plt.grid(True)
plt.show()
```

**Bài 3. Error Chart (Biểu đồ sai số)**

**Đề bài:** Viết chương trình hiển thị biểu đồ cột với sai số ±5% trên mỗi cột, thể hiện giá trị trung bình điểm kiểm tra của các lớp học.

**Dữ liệu:**

classes = ['Class A', 'Class B', 'Class C']

average_scores = [75, 82, 90]

errors = [4, 5, 3]


```python
# prompt: Viết chương trình hiển thị biểu đồ cột với sai số ±5% trên mỗi cột, thể hiện giá trị trung bình điểm kiểm tra của các lớp học.

import matplotlib.pyplot as plt

# Dữ liệu
classes = ['Class A', 'Class B', 'Class C']
average_scores = [75, 82, 90]
errors = [4, 5, 3]

# Tính toán sai số trên và dưới
upper_errors = [score + error for score, error in zip(average_scores, errors)]
lower_errors = [score - error for score, error in zip(average_scores, errors)]

# Vẽ biểu đồ cột với sai số
plt.figure(figsize=(8, 6))
plt.bar(classes, average_scores, yerr=[
         [score - lower for score, lower in zip(average_scores, lower_errors)],
         [upper - score for upper, score in zip(upper_errors, average_scores)]
         ],
        capsize=5, color='skyblue', edgecolor='black', linewidth=1.2)

# Thêm tiêu đề và nhãn
plt.title('Biểu đồ điểm trung bình các lớp với sai số')
plt.xlabel('Lớp học')
plt.ylabel('Điểm trung bình')

# Hiển thị biểu đồ
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

**Bài 4. Heatmap (Bản đồ nhiệt)**

**Đề bài:** Tạo bản đồ nhiệt thể hiện điểm kiểm tra của học sinh trong các môn học.

**Dữ liệu:**

import numpy as np

students = ['Anna', 'Ben', 'Chris']

subjects = ['Math', 'English', 'Science']

scores = np.array([[90, 85, 88], [78, 92, 84], [85, 80, 86]])


```python
# prompt: Tạo bản đồ nhiệt thể hiện điểm kiểm tra của học sinh trong các môn học.

import matplotlib.pyplot as plt
import numpy as np

# Dữ liệu
students = ['Anna', 'Ben', 'Chris']
subjects = ['Math', 'English', 'Science']
scores = np.array([[90, 85, 88], [78, 92, 84], [85, 80, 86]])

# Tạo bản đồ nhiệt
plt.imshow(scores, cmap='viridis', interpolation='nearest')

# Thêm nhãn cho trục
plt.xticks(np.arange(len(subjects)), subjects)
plt.yticks(np.arange(len(students)), students)

# Thêm thanh màu
plt.colorbar(label='Điểm số')

# Hiển thị điểm số trên bản đồ nhiệt
for i in range(len(students)):
    for j in range(len(subjects)):
        plt.text(j, i, scores[i, j], ha='center', va='center', color='w')

# Thêm tiêu đề
plt.title('Bản đồ nhiệt điểm kiểm tra')

# Hiển thị biểu đồ
plt.show()
```

**Bài 5. Gantt Chart (Biểu đồ Gantt**)

**Đề bài:** Tạo biểu đồ Gantt thể hiện tiến độ thực hiện các giai đoạn của một dự án học sinh.

**Dữ liệu:**

tasks =
[
    dict(Task="Nghiên cứu", Start='2025-04-01', Finish='2025-04-05'),

    dict(Task="Phân tích", Start='2025-04-06', Finish='2025-04-10'),

    dict(Task="Viết báo cáo", Start='2025-04-11', Finish='2025-04-15')
]


```python
# prompt: Tạo biểu đồ Gantt thể hiện tiến độ thực hiện các giai đoạn của một dự án học sinh.

import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime, timedelta

# Dữ liệu
tasks = [
    dict(Task="Nghiên cứu", Start='2025-04-01', Finish='2025-04-05'),
    dict(Task="Phân tích", Start='2025-04-06', Finish='2025-04-10'),
    dict(Task="Viết báo cáo", Start='2025-04-11', Finish='2025-04-15')
]

# Chuyển đổi chuỗi ngày tháng thành đối tượng datetime
for task in tasks:
    task['Start'] = datetime.strptime(task['Start'], '%Y-%m-%d')
    task['Finish'] = datetime.strptime(task['Finish'], '%Y-%m-%d')

# Tạo biểu đồ Gantt
fig, ax = plt.subplots(figsize=(10, 6))

for task in tasks:
    duration = task['Finish'] - task['Start']
    ax.barh(task['Task'], duration.days, left=task['Start'], height=0.5, align='center', edgecolor='black')

# Định dạng trục thời gian
ax.xaxis_date()
date_format = mdates.DateFormatter('%Y-%m-%d')
ax.xaxis.set_major_formatter(date_format)
fig.autofmt_xdate()

# Thêm tiêu đề và nhãn
ax.set_title('Biểu đồ Gantt Tiến độ Dự án')
ax.set_xlabel('Thời gian')
ax.set_ylabel('Công việc')
ax.grid(True, axis='x', linestyle='--', alpha=0.7)

# Hiển thị biểu đồ
plt.tight_layout()
plt.show()
```

**Bài 6. Circular Gauge (Đồng hồ đo tròn)**

**Đề bài:** Viết chương trình thể hiện mức độ hoàn thành công việc dưới dạng đồng hồ đo tròn (gauge).

**Dữ liệu:**

completion = 72  # phần trăm hoàn thành


```python
import matplotlib.pyplot as plt
import numpy as np

# Dữ liệu đầu vào
completion = 72  # phần trăm hoàn thành

# Tạo biểu đồ với hệ tọa độ polar (tròn)
fig, ax = plt.subplots(figsize=(8, 5), subplot_kw={'projection': 'polar'})

# Chuyển completion thành radian (0–π)
theta = (completion / 100) * np.pi

# Vẽ nền nửa vòng tròn (xám)
ax.barh(1, np.pi, left=0, height=0.4, color='#dddddd')

# Vẽ phần hoàn thành (màu xanh lá)
ax.barh(1, theta, left=0, height=0.4, color='#00cc66')

# Vẽ kim chỉ mức độ hoàn thành (màu đỏ)
ax.plot([theta, theta], [0, 1.1], color='red', linewidth=2)

# Chỉnh hướng bắt đầu từ trái sang phải
ax.set_theta_zero_location("W")
ax.set_theta_direction(-1)

# Ẩn trục
ax.set_xticks([])
ax.set_yticks([])
ax.spines['polar'].set_visible(False)

# Thêm số % bằng hệ Cartesian, không dùng trục polar
fig.text(0.5, 0.15, f"{completion}%", ha='center', va='center',
         fontsize=28, fontweight='bold', color='black')

# Tiêu đề
plt.title("Đồng hồ đo nửa hình tròn", fontsize=16, pad=20)
plt.tight_layout()
plt.show()
```
