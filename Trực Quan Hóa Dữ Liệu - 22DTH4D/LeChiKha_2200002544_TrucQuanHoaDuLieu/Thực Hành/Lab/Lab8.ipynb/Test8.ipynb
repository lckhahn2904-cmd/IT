**Bài 4: Vẽ biểu đồ với trục Y theo tỷ lệ phần trăm trong Matplotlib:**

Để vẽ một biểu đồ với trục Y dưới dạng phần trăm trong matplotlib, chúng ta có thể thực
hiện các bước sau:

* Tạo một danh sách các số như y.
* Tạo một số ngăn.
* Vẽ một biểu đồ sử dụng phương thức hist(), trong đó y, bins và edgecolor được chuyển vào đối số. Lưu trữ các bản vá để đặt tỷ lệ phần trăm trên trục Y.
* Tạo một danh sách các màu từ các số chữ và số đã cho.
* Để đặt tỷ lệ phần trăm, hãy lặp lại các bản vá (thu được ở bước 3).
* Đặt phạm vi đánh dấu trục Y.
* Để hiển thị hình, sử dụng phương thức show().

***Ví dụ:***

```python
import matplotlib.pyplot as plt
import numpy as np
import random
plt.rcParams["figure.figsize"] = [7.00, 3.50]
plt.rcParams["figure.autolayout"] = True
y = [4, 1, 8, 7, 6, 4, 2, 1, 2, 5]
nbins = 10
_,_, patches = plt.hist(y, bins=nbins, edgecolor='white')
colors = ["#" + ''.join([random.choice('0123456789ABCDEF') for j in range(6)])
for i in range(nbins)]
for patch in patches:
    patch.set_facecolor(colors[np.random.randint(100) % nbins])
    x = patch.get_x() + patch.get_width() / 2
    y = patch.get_height() + .05
    plt.annotate('{:.1f}%'.format(y), (x, y), ha='center')
plt.ylim(0, 3)
plt.show()
```
