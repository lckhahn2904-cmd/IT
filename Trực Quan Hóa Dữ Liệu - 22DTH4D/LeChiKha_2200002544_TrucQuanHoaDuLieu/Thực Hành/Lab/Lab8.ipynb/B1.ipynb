Hãy vẽ biểu đồ Histogram với dữ liệu trên theo kết quả: (cho edgecolor: biên màu, color:màu sắc)

```python
import matplotlib.pyplot as plt

x = [87, 53, 66, 61, 67, 68, 62, 110,
     104, 61, 111, 123, 117, 119, 116,
     104, 92, 111, 90, 103, 81, 80, 101,
     51, 79, 107, 110, 129, 145, 128,
     132, 135, 131, 126, 139, 110]

plt.hist(x, color='green', edgecolor='red')
plt.show()
```
