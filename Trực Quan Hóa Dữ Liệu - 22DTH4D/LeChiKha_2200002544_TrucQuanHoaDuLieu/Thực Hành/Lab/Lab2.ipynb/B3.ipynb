**Bài 3. Vẽ biểu đồ line cho list các điểm:**

       [0,1,4,6,5], [1,2,3,5,6]

       [0,1,2,7,8], [10,4,3,2,1]

       [2.5,2.5,2.5,1.5,0.5], [1,3,5,7,10]

    Tạo 3 ax là 3 biểu đồ:

```python
import matplotlib.pyplot as plt

# Dữ liệu các đường
x1, y1 = [0,1,4,6,5], [1, 2, 3, 5, 6]
x2, y2 = [0,1,2,7,8], [10, 4, 3, 2, 1]
x3, y3 = [2.5, 2.5, 2.5, 1.5, 0.5], [1, 3, 5, 7, 10]

# Tạo figure với 3 subplot, cùng chia sẻ trục Y
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(12, 4), sharey=True, dpi=120)

# Vẽ từng đường trên từng subplot với màu sắc và marker phù hợp
ax1.plot(x1, y1, 'g-', marker='o')
ax1.set_title("green color")
ax1.set_xlabel("x")
ax1.set_ylabel("y")

ax2.plot(x2, y2, 'r-', marker='o')
ax2.set_title("red color")
ax2.set_xlabel("x")

ax3.plot(x3, y3, 'b-', marker='o')
ax3.set_title("blue color")
ax3.set_xlabel("x")

# Hiển thị lưới trên tất cả subplot
for ax in [ax1, ax2, ax3]:
    ax.grid(True)

# Sắp xếp layout hợp lý
plt.tight_layout()
plt.show()
```
