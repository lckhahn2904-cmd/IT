**Bài 1: Cách sử dụng hàm plot để vẽ biểu đồ.**
     # Vẽ x và y sử dụng kiểu đường thẳng và màu sắc mặc định.
     # Vẽ x và y sử dụng kiểu loại của điểm Markers là hình tròn.
     # Vẽ y sử dụng x như một mảng index mặc định : 0,1,2...N-1. Nhưng
kiểu loại là dấu '+'
Cho y = f(x) trong biểu đồ bởi hàm plot(x, y).  
Ví dụ vẽ biểu đồ từ điểm có tọa độ (0,0) tới điểm có tọa độ (5,200) như sau:

```python
import matplotlib.pyplot as plt
import numpy as np

x_points = np.array([0,5])
y_points = np.array([0,200])

plt.plot(x_points, y_points)
plt.show()
```

**Bài 2: Vẽ các điểm và không vẽ đường thẳng trong biểu đồ trong python: Hàm này được dùng cho trường hợp chỉ vẽ các điểm, không vẽ đường thẳng trong biểu đồ. Tham số 'o' là các điểm được biểu diễn bởi các hình tròn.**

```python
import matplotlib.pyplot as plt
import numpy as np

x_points = np.array([1,10])
y_points = np.array([0,100])

plt.plot(x_points, y_points, 'o')
plt.show()
```

**Bài 3: Vẽ đồ thị định dạng Plot:**

Phương thức axis () cho phép đặt chế độ xem cho các trục theo xmin, xmax, ymin và ymax.  
Chuỗi định dạng ‘gs’ ở đây cho chúng ta các ô vuông màu xanh lá cây. Tương tự, ta cũng
có thể vẽ các quả cầu màu xanh lá cây, đỏ và xanh lam, hình vuông và hình tam giác.  

```python
import matplotlib.pyplot as plt
import numpy as np

t=np.arange(0,5,0.2)
# vẽ đồ thị
plt.plot(t,t, 'r--',t,t**3, 'b^',t,t**2, 'gs')
plt.show()
```
