**Bài 2 .Vẽ biểu đồ line cho list các điểm:**

       [0,1,2,3,4], [1,2,3,4,10]

       [0,1,2,3,4], [10,4,3,2,1]

       [2.5,2.5,2.5,1.5,0.5], [1,3,5,7,10]

       [0,1.4,2,5,6], [3,2,3,5,8]
       

```python
import matplotlib.pyplot as plt
import numpy as np
# Data
x1, y1 = [0, 1, 2, 3, 4], [1, 2, 3, 4, 10]
x2, y2 = [0, 1, 2, 3, 4], [10, 4, 3, 2, 1]
x3, y3 = [2.5, 2.5, 2.5, 1.5, 0.5], [1, 3, 5, 7, 10]
x4, y4 = [0, 1.4, 2, 5, 6], [3, 2, 3, 5, 8]

# Vẽ các đường với màu sắc khác nhau
plt.figure(figsize=(10, 4))
plt.plot(x1, y1, 'g-', marker='o', label='green color')
plt.plot(x2, y2, 'r-', marker='o', label='red color')
plt.plot(x3, y3, 'b-', marker='o', label='blue color')
plt.plot(x4, y4, 'y-', marker='o', label='yellow color')

plt.gca().set_facecolor('lightgray')
plt.title("Line in Python Matplotlib")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```
