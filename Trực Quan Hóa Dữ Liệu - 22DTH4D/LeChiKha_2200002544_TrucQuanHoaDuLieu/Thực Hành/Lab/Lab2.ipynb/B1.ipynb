**Bài 1. Vẽ biểu đồ hình sin cho thư viện numpy (np.linspace((-20,20,80), độ dày =4:**

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-20, 20, 80)
y = np.sin(x)

# Vẽ biểu đồ
plt.figure(figsize=(10, 4))
plt.plot(x, y, 'r--',linewidth=4)

# Tùy chỉnh nền
plt.gca().set_facecolor('lightgray')

plt.title("Hàm Sin")
plt.xlabel("X value")
plt.ylabel("Sin of x")
plt.show()
```
