***Ví dụ 1: Trực quan hóa các nốt theo x và y***

```python
# Ví dụ 1
import matplotlib.pyplot as plt
import numpy as np
# data hiển thị trên plots
x = np.array ([3, 1, 3, 12, 2, 4, 4, 2, 8, 7, 9, 16, 7, 5, 9, 10])
y = np.array ([3, 2, 1, 4, 5, 6, 7, 2, 4, 4, 9, 13, 15, 8, 9, 7 ])

# Đồ thị đơn giản scatter chart
plt.scatter(x, y)

# Title
plt.title("Scatter simple chart")
plt.show()
```

***Ví dụ 2: Tạo hàm ngẫu nhiên với 75 nốt với màu sắc ngẫu nhiên hiển thị alpha=0.5. Vẽ biểu đồ thể hiện scatter cho các nốt***

```python
# Ví dụ 2
import numpy as np
import matplotlib.pyplot as plt

# hàm khởi tạo các bộ sinh số ngẫu nhiên (random generator).
# Biển số trong seed thường là một số nguyên không âm 32 bit.
np.random.seed(13499202)

# Gán số lượng N (Bong bóng)
N = 75
x = np.random.rand(N)
y = np.random.rand(N)

# Gán màu sắc random
colors = np.random.rand(N)

# Gán bán kính random cho các điểm từ 0 đến 40 (độ lớn Bình Phương)
area = (40 * np.random.rand(N))**2

# in theo x và y (diện tích và màu sắc)
plt.scatter(x, y, s=area, c=colors, alpha=0.5)
plt.show()
```

***Ví dụ 3: Cho các thuộc tính:***

Màu c ="yellow",

Đọ dày của viền linewidths = 2,

Điểm kiểu vuông marker ="s",

Biên của điểm edgecolor ="green",

Các điểm có kích thước 100

Hãy vẽ biểu đồ scatter các điểm?

```python
# Ví dụ 3
import matplotlib.pyplot as plt

# dataset của x và y
x = [89, 43, 36, 36, 95, 10, 66, 34, 38, 20, 46, 3, 35]
y = [21, 46, 3, 35, 67, 95, 53, 72, 58, 10, 36, 36, 95]

# Tạo các nốt với các thuộc tính sau:
plt.scatter(x, y, c ="yellow",
            linewidths = 2,
            marker ="s",
            edgecolor ="green",
            s = 100)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```
