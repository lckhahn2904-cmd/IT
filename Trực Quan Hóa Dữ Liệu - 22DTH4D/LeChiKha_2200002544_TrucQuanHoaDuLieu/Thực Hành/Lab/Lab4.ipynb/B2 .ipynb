**BÀI TẬP 2: Vẽ biểu đồ scatter như biểu đồ sau, có 4 màu với các kiểu khác nhau:**

Cho n1, n2, n3, n4: với random(20)

```python
import matplotlib.pyplot as plt
import numpy as np

# Số lượng điểm cho mỗi nhóm
n1, n2, n3, n4 = 20, 20, 20, 20

# Tạo dữ liệu ngẫu nhiên
x1, y1 = np.random.rand(n1), np.random.rand(n1)
x2, y2 = np.random.rand(n2), np.random.rand(n2)
x3, y3 = np.random.rand(n3), np.random.rand(n3)
x4, y4 = np.random.rand(n4), np.random.rand(n4)

# Vẽ scatter plot
plt.scatter(x1, y1, color='blue', marker='^', label='blue')
plt.scatter(x2, y2, color='red', marker='v', label='red')
plt.scatter(x3, y3, color='green', marker='*', label='green')
plt.scatter(x4, y4, color='orange', marker='o', label='orange')

plt.legend()
plt.show()

```
