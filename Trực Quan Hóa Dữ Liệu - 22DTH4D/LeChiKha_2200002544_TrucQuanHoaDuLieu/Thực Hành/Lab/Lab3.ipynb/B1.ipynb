Thông thường, một biểu đồ phải có chú thích của trục x và trục y theo đối tượng mình cần trực quan hóa

Ta có thể thêm Title và Axis Labels với các số cuộn trên trục x và các dấu trên trục y.

Trực quan hóa doanh thu (Triệu đồng) của các Cửa hàng trong tháng

```python
import matplotlib.pyplot as plt
import numpy as np

marks=[72,40,33,90,100]
bars=('Cửa hàng 1', 'Cửa hàng 2', 'Cửa hàng 3', 'Cửa hàng 4', 'Cửa hàng 5')
y=np.arange(len(bars))

plt.bar(bars,marks,color='y')
plt.xticks(y,bars)
plt.title('Doanh thu bán hàng trong tháng')
plt.xlabel('Số lượng Cửa hàng')
plt.ylabel('Doanh thu đạt được (Triệu đồng)')
plt.show()
```
