- Hãy trực quan hóa dữ liệu Bar chart hiển thị số ngày trong tuần có được tiền thưởng (tip)

```python
import matplotlib.pyplot as plt
import numpy as np

marks = [6.4, 5, 6.6, 3, 10, 6.5, 4.3]
bars = ['Sun', 'Tue', 'Wed', 'Mon', 'Sat', 'Thur', 'Fri']
y = np.arange(len(bars))

plt.figure(figsize=(8, 6))
plt.bar(bars, marks, color='blue')
plt.xticks(rotation=0)
plt.title('Hiển thị số tiền thưởng trong ngày ($)')
plt.xlabel('Ngày trong tuần')
plt.ylabel('Tip (Tiền thưởng)')
plt.show()
```
