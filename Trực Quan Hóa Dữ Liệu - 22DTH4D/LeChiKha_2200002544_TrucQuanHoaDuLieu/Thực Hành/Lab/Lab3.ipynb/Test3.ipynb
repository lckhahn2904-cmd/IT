**A. Vẽ biểu đồ cột (Matplotlib Bar).**

Biểu đồ thanh trong Python, còn được gọi là biểu đồ thanh, biểu thị cách một biến số liên quan đến một biến phân loại.

***Ví dụ 1: Vertical Bar Plot (Biểu đồ thanh dọc)***  


```python
import matplotlib.pyplot as plt
import numpy as np

marks=[72,40,33,90,100]
bars=('Roll 1', 'Roll 2', 'Roll 3', 'Roll 4', 'Roll 5')
y=np.arange(len(bars))
plt.bar(y,marks,color='r')
```

***Ví dụ 2: Horizontal Bar Plot ( Biểu đồ thanh ngang)***

```python
import matplotlib.pyplot as plt
import numpy as np

marks=[72,40,33,90,100]
bars=('Roll 1', 'Roll 2', 'Roll 3', 'Roll 4', 'Roll 5')
y=np.arange(len(bars))
plt.barh(y,marks,color='b')
plt.yticks(y,bars)
```

***Ví dụ 3: Bar thể hiện Nhóm tuổi của Male và Female***

```python
import matplotlib.pyplot as plt
import numpy as np

Male = [30, 34, 37, 35, 27]
Female = [25, 32, 34, 40, 25]
group = ['Nhóm 1', 'Nhóm 2', 'Nhóm 3', 'Nhóm 4', 'Nhóm 5']

# Lấy cột từ group
x = np.arange(len(group))
# độ rộng của bars
width = 0.35

# Vẽ biểu đồ cột
fig, ax = plt.subplots()
rects1 = ax.bar(x - width/2, Male, width, color='g', label='Male')
rects2 = ax.bar(x + width/2, Female, width, color='b', label='Female')

# Gán Label và Title
ax.legend()
ax.set_ylabel('Độ tuổi')
ax.set_title('Nhóm tuổi và Giới tính')
ax.set_xticks(x)
ax.set_xticklabels(group)

```
