**Bài 2. Kết hợp vẽ 2 Lines để hiển thị kết quả sau:**

```python
import matplotlib.pyplot as plt
import numpy as np

# Đỉnh trục  x
x_points = np.array([1,3,6,9,15])
x1_points = np.array([1,3,7,13])

# Đỉnh trục  y
y_points = np.array([10,9,22,15,50])
y1_points = np.array([9,22,15,45])

plt.plot(x_points, y_points, linestyle=':', color='g')
plt.plot(x1_points, y1_points, linestyle='--', color='r')

plt.show()
plt.show()
```
