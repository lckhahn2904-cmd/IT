```python
import numpy as np
import matplotlib.pyplot as plt

# Đỉnh trục  x
x_points = np.array([20, 22, 24, 26, 28, 30])

# Đỉnh trục  y
y_points = np.array([5, 3, 5, 4, 5, 5])
y1_points = np.array([4, 4.5, 5, 3, 4, 7])
y2_points = np.array([5, 4, 6, 4, 6, 8])

plt.plot(x_points, y_points, linestyle='-', color='b', label='Xăng')
plt.plot(x_points, y1_points, linestyle='-', color='g',  label='Dầu')
plt.plot(x_points, y2_points, linestyle='-', color='r', label='Nhớt')


plt.xlabel('Giá nhiên liệu (VNĐ)')
plt.ylabel('Thời Gian (Tháng)')
plt.title('Biểu đồ nhiên liệu trong nước (năm 2020)')
plt.legend()
plt.show()
```
