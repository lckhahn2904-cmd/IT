**Vẽ biểu đồ Line trong python có dạng nét đứt bởi các dấu chấm (dotted)**

```python
import matplotlib.pyplot as plt
import numpy as np
# Đỉnh trục x
x_points = np.array([1, 3, 5, 7])
# Đỉnh trục y
y_points = np.array([15, 25, 20, 40])
plt.plot(x_points, y_points, linestyle = 'dotted') # style line
plt.show()
```

**Vẽ biểu đồ đường Line trong python có dạng nét đứt bởi các dấu gạch (dashed)**

```python
import matplotlib.pyplot as plt
import numpy as np
x_points = np.array([1, 3, 8, 15])
y_points = np.array([10, 20, 15, 45])
plt.plot(x_points, y_points, linestyle = 'dashed')
plt.show()
```

**Cách thay đổi màu sắc của đường (Line) trong biểu đồ (Line Color)**
* Sử dụng từ khóa color hoặc c để thiết lập biểu thị màu sắc của đường.  
*  Tham số màu :  r - red, g -green, b - blue, y - yellow, … Vẽ biểu đồ đường nét đứt dashed  và có màu đỏ.


```python
import matplotlib.pyplot as plt
import numpy as np
x_points = np.array([1, 3, 7, 13])
y_points = np.array([9, 23, 15, 45])
plt.plot(x_points, y_points, linestyle='dashed', color = 'r')
plt.show()
```

**Cách vẽ có định dạng Đỉnh**
-	Định dạng đường:  - (đường thẳng), -- (đường nét đứt), nếu không truyền tham số là không vẽ đường.
-	Định dạng đỉnh:  * (đỉnh hình bông tuyết), D (đỉnh hình thoi), s (đỉnh hình vuông), go (chấm tròn), ...

**Ví dụ: Thêm tham số b tức là màu blue, D tức là đỉnh hình thoi và -- tức là cạnh nét đứt.**


```python
import matplotlib.pyplot as plt
import numpy as np
x_points = np.array([1, 2, 3, 5, 7])
y_points = np.array([5, 2, 8, 12, 20])
plt.plot(x_points, y_points,'bD--')
plt.show()
```
