**BÀI 1: Vẽ Lines and Points 3-D:**

Đồ thị có đường và điểm là đồ thị 3 chiều đơn giản nhất. ax.plot3d và ax.scatter lần lượt
là hàm vẽ đồ thị đường và điểm.

***Ví dụ: Đồ thị đường 3 chiều***

```python
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure()

# Cú pháp cho phép chiếu 3-D
ax = plt.axes(projection='3d')

# Xác định cả 3 trục
z = np.linspace(0, 1, 100)
x = z * np.sin(25 * z)
y = z * np.cos(25 * z)

# Plotting
ax.plot3D(x, y, z, 'green')
ax.set_title('3D line plot')
plt.show()
```

***Ví dụ: 3 dimensional scattered graph***

```python
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure()

# Cú pháp cho phép chiếu 3-D
ax = plt.axes(projection='3d')

# Xác định trục
z = np.linspace(0, 1, 100)
x = z * np.sin(25 * z)
y = z * np.cos(25 * z)
c = x + y
ax.scatter3D(x, y, z, c=c)

# Plotting
ax.set_title('3D Scatter plot')
plt.show()
```

**BÀI 2: Vẽ đồ thị Surface và Wireframes:**

Biểu đồ bề mặt và biểu đồ Wireframes hoạt động trên dữ liệu dạng lưới.

Lấy giá trị lưới và vẽ trên bề mặt ba chiều.

***Ví dụ 1: Surface graph***

```python
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt

# Xác định bề mặt và trục
x = np.outer(np.linspace(-2, 2, 10), np.ones(10))
y = x.copy().T
z = np.cos(x ** 2 + y ** 3)

fig = plt.figure()

# Cú pháp cho 3-D plotting
ax = plt.axes(projection='3d')

# Plotting
ax.plot_surface(x, y, z, cmap ='viridis', edgecolor ='green')
ax.set_title('Surface plot geeks for geeks')
plt.show()
```

***Ví dụ 2: Wireframes***

```python
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt

# Hàm của trục z
def f(x, y):
    return np.sin(np.sqrt(x ** 2 + y ** 2))

# Xác định trục
x = np.linspace(-1, 5, 10)
y = np.linspace(-1, 5, 10)

X,Y = np.meshgrid(x, y)
Z = f(X, Y)

fig = plt.figure()
ax = plt.axes(projection='3d')
ax.plot_wireframe(X, Y, Z, color ='red')
ax.set_title('Wireframe')
```

**BÀI 3: Vẽ dải Möbius:**

Dải Mobius còn được gọi là hình trụ xoắn, là một bề mặt- một mặt không có ranh giới. Để tạo dải Mobius, ta có tham số hóa của nó, đó là dải hai chiều và cần hai chiều nội tại.

Phạm vi góc của nó từ 0 đến 2 hình tròn quanh vòng và chiều rộng nằm trong khoảng từ -1 đến 1.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Xác định các tham số của Möbius
R = 2

# Định nghĩa bề mặt Möbius strip
u = np.linspace(0, 2 * np.pi, 100)
v = np.linspace(-1, 1, 10)
u, v = np.meshgrid(u, v)
x = (R + v * np.cos(u/2)) * np.cos(u)
y = (R + v * np.cos(u/2)) * np.sin(u)
z = v * np.sin(u/2)

# Tạo plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Bề mặt của Plot the Möbius strip
ax.plot_surface(x, y, z, alpha=0.5)

# Set thuộc tính plot
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Möbius Strip')

# Set giới hạn plot
ax.set_xlim([-3, 3])
ax.set_ylim([-3, 3])
ax.set_zlim([-3, 3])

# Hiển thị plot
plt.show()
```
