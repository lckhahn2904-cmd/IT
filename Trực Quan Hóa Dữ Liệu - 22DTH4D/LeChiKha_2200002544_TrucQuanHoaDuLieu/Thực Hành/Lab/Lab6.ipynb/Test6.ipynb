Biểu đồ hộp (còn gọi là biểu đồ nén và biểu đồ râu) sử dụng các hộp và đường để mô tả sự phân bố của một hoặc nhiều nhóm dữ liệu số.  

Giới hạn hộp cho biết phạm vi của 50% dữ liệu trung tâm, với một đường trung tâm đánh dấu giá trị trung bình. Các dòng mở rộng từ mỗi hộp để nắm bắt phạm vi của dữ liệu còn lại, với các dấu chấm được đặt qua các cạnh của dòng để biểu thị các giá trị ngoại lệ.

Biểu đồ hiển thị tóm tắt tập hợp dữ liệu có chứa minimum, phần tư thứ nhất, trung vị, phần tư thứ ba và maximum.  

Sử dụng hàm numpy.random.normal () để tạo dữ liệu giả.  
Cần ba đối số, giá trị trung bình, độ lệch chuẩn của phân phối chuẩn và số lượng giá trị mong muốn.

**Bài 1: Tạo Danh sách các mảng là đầu vào bắt buộc để tạo boxplot.**

```python
import numpy as np
import matplotlib.pyplot as plt

# Tạo mảng ngẫu nhiên
np.random.seed(10)
collech_1 = np.random.normal(100, 10, 200)
collech_2 = np.random.normal(80, 30, 200)
collech_3 = np.random.normal(90, 20, 200)
collech_4 = np.random.normal(70, 25, 200)

fig = plt.figure()

# Tạo 1 ax
ax = fig.add_axes([0,0,1,1])

# Tạo boxplot
bp = ax.boxplot([collech_1, collech_2, collech_3, collech_4])
plt.show()
```

**Bài 2: Biểu đồ Box plot cơ bản:**

```python
import numpy as np
import matplotlib.pyplot as plt

# Trạng thái ngẫu nhiên có khả năng tái tạo
np.random.seed(19680801)

# Tạo dữ liệu ngẫu nhiên
spread = np.random.rand(50) * 100
center = np.ones(25) * 50
flier_high = np.random.rand(10) * 100 + 100
flier_low = np.random.rand(10) * -100

data = np.concatenate((spread, center, flier_high, flier_low))
fig1, ax1 = plt.subplots()
ax1.set_title('Basic Plot')
ax1.boxplot(data)
plt.show()
```

**Bài 3:  Sử dụng hàm boxplot() của seaborn. Nhìn vào biểu đồ, sẽ dễ dàng kết luận rằng nhóm ‘C’ có giá trị cao hơn các nhóm khác. Tuy nhiên, ta không thể thấy sự phân bố cơ bản của các dấu chấm trong mỗi nhóm, cũng như số lượng quan sát cho mỗi nhóm.**

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import pandas as pd

# Data
a = pd.DataFrame({'group': np.repeat('A', 500), 'value': np.random.normal(10, 5, 500)})
b = pd.DataFrame({'group': np.repeat('B', 500), 'value': np.random.normal(13, 1.2, 500)})
c = pd.DataFrame({'group': np.repeat('B', 500), 'value': np.random.normal(18, 1.2, 500)})
d = pd.DataFrame({'group': np.repeat('C', 20), 'value': np.random.normal(25, 4, 20)})
e = pd.DataFrame({'group': np.repeat('D', 100), 'value': np.random.uniform(12, size=100)})

# Gộp dữ liệu
df=a.append(b).append(c).append(d).append(e)

# Biểu diễn boxplot
sns.boxplot(x='group', y='value', data=df)
plt.show()
```
