```python
import requests
import pandas as pd
from datetime import datetime, timedelta

# toạ độ trung tâm Hải Phòng
lat = 20.87
lon = 106.68
start = "20100101"
end   = "20241231"

# Danh sách thông số muốn lấy (kiểm tra xem POWER hỗ trợ trong daily)
parameters = [
    "T2M",        # nhiệt độ trung bình 2 m
    "T2M_MAX",    # nhiệt độ cao nhất
    "T2M_MIN",    # nhiệt độ thấp nhất
    "PRECTOTCORR",# lượng mưa (mm/day)
    "RH2M",       # độ ẩm tương đối
    "WS10M",      # tốc độ gió 10m
    "WD10M",      # hướng gió 10m  — bạn kiểm tra xem alias có WD10M hay WD2M
    "PS",         # áp suất bề mặt
    "T2MDEW",     # điểm sương
    "ALLSKY_SFC_SW_DWN"  # bức xạ ngắn tới mặt đất (MJ/m²/day)
    ]

# Chuyển danh sách thành chuỗi cho API
par_str = ",".join(parameters)
base_url = "https://power.larc.nasa.gov/api/temporal/daily/point"
params = {
    "start": start,
    "end": end,
    "latitude": lat,
    "longitude": lon,
    "parameters": par_str,
    "community": "AG",   # cộng đồng “agroclimatology” phù hợp dữ liệu khí + cải vĩ
    "format": "CSV"
    }

print("Requesting:", base_url, params)
resp = requests.get(base_url, params=params)
resp.raise_for_status()

# Lưu CSV
fname = "HaiPhong_power_daily_2010_2024.csv"
with open(fname, "wb") as f:
    f.write(resp.content)
print("Saved:", fname)

# Define column names based on the file header
column_names = ['YEAR', 'DOY', 'T2M', 'T2M_MAX', 'T2M_MIN', 'PRECTOTCORR', 'RH2M', 'WS10M', 'WD10M', 'PS', 'T2MDEW', 'ALLSKY_SFC_SW_DWN']
# Read data into DataFrame, skipping initial metadata rows and providing column names
df = pd.read_csv(fname, skiprows=17, names=column_names)
print(df.head())
```
