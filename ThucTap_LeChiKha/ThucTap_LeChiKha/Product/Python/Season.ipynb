```python
import pandas as pd
# Danh sách tỉnh theo vùng
mien_bac = [ "Hà Nội", "Hải Phòng", "Quảng Ninh", "Bắc Ninh", "Bắc Giang", "Hưng Yên", "Hải Dương", "Hà Nam", "Nam Định", "Ninh Bình", "Thái Bình", "Vĩnh Phúc", "Phú Thọ", "Hòa Bình", "Sơn La", "Điện Biên", "Lai Châu", "Lào Cai", "Yên Bái", "Hà Giang", "Tuyên Quang", "Cao Bằng", "Bắc Kạn", "Thái Nguyên", "Lạng Sơn" ]
mien_trung = [ "Thanh Hóa", "Nghệ An", "Hà Tĩnh", "Quảng Bình", "Quảng Trị", "Thừa Thiên Huế", "Đà Nẵng", "Quảng Nam", "Quảng Ngãi", "Bình Định", "Phú Yên", "Khánh Hòa", "Ninh Thuận", "Bình Thuận", "Kon Tum", "Gia Lai", "Đắk Lắk", "Đắk Nông", "Lâm Đồng" ]
mien_nam = [ "Hồ Chí Minh", "Bình Dương", "Đồng Nai", "Bà Rịa - Vũng Tàu", "Bình Phước", "Tây Ninh", "Long An", "Tiền Giang", "Bến Tre", "Vĩnh Long", "Trà Vinh", "Đồng Tháp", "An Giang", "Kiên Giang", "Cần Thơ", "Hậu Giang", "Sóc Trăng", "Bạc Liêu", "Cà Mau" ]
tat_ca_tinh = mien_bac + mien_trung + mien_nam
# Hàm xác định mùa theo vùng
def xac_dinh_mua_theo_vung(thang, tinh):
    if tinh in mien_bac:
        if thang in [2, 3, 4]:
            return "Xuân"
        elif thang in [5, 6, 7, 8]:
            return "Hè"
        elif thang in [9, 10]:
            return "Thu"
        else:
            return "Đông"

    elif tinh in mien_trung:
        if thang in [1, 2, 3, 4, 5, 6, 7, 8]:
            return "Mùa khô"
        else:
            return "Mùa mưa"

    elif tinh in mien_nam:
        if thang in [11, 12, 1, 2, 3, 4]:
            return "Mùa khô"
        else:
            return "Mùa mưa"
    return "Không xác định"

# Tạo dữ liệu thời gian
dates = pd.date_range(start="2010-01-01", end="2024-12-31", freq="D")
df = pd.DataFrame({"Ngày": dates})
df["Năm"] = df["Ngày"].dt.year
df["Tháng"] = df["Ngày"].dt.month
# Nhân bản cho các tỉnh
df_full = pd.concat([df.assign(Tỉnh=t) for t in tat_ca_tinh], ignore_index=True)
# Xác định mùa
df_full["Mùa"] = df_full.apply(
    lambda row: xac_dinh_mua_theo_vung(row["Tháng"], row["Tỉnh"]),
    axis=1
)
# Sắp xếp theo Ngày
df_full = df_full.sort_values(by="Ngày")
# Định dạng ngày dd/mm/yyyy khi xuất CSV
df_full["Ngày"] = df_full["Ngày"].dt.strftime("%d/%m/%Y")
# Xuất file
output_file = "mua_vietnam_2010_2024.csv"
df_full.to_csv(output_file, index=False, encoding="utf-8-sig")
print("Đã tạo file:", output_file)
print(df_full.head())
```
