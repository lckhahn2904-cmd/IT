```python
df = pd.read_csv('/content/earthquake_1995-2023.csv')
print(df.columns.tolist())
df[0:2].T
```

```python
print(df['year'].min(),df['year'].max())
print(df['magnitude'].min(),df['magnitude'].max())
print(df['depth'].min(),df['depth'].max())
```
